```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
```

```python
urlcnn = 'https://www.cnnindonesia.com/search?query=politik&kanal=nasional&fromdate=01%2F01%2F2025&todate=31%2F01%2F2025'
rescnn = requests.get(urlcnn, headers=headers)

soupcnn = BeautifulSoup(rescnn.text, 'lxml')
# print(soup)
```

```python
# <a class="text-cnn_black dark:text-white inline-flex items-center justify-center w-[30px] h-[30px] rounded-md hover:bg-cnn_grey dark:hover:bg-cnn_black" href="/search?query=politik&amp;kanal=nasional&amp;fromdate=01/01/2025&amp;todate=31/01/2025&amp;page=15" dtr-evt="search result" dtr-sec="search result" dtr-act="halaman search" dtr-idx="15" collectedonscroll="true">15</a>

# Setup driver sele
driver = webdriver.Chrome()
driver.get(urlcnn)

try:
    wait = WebDriverWait(driver, 10)
    pagination_container_selector = 'div.flex.items-center.justify-center.gap-4'
    wait.until(EC.visibility_of_element_located((By.CSS_SELECTOR, pagination_container_selector)))

    print("masukk")

    # copy html
    html_content = driver.page_source
    soup = BeautifulSoup(html_content, 'lxml')

    # select bsoup
    container = soup.select_one(pagination_container_selector)
    
    if container:
        all_links = container.find_all('a')
        if all_links:
            
            page_numbers = [link.get_text(strip=True) for link in all_links if link.get_text(strip=True).isdigit()]
            if page_numbers:
                last_page = page_numbers[-1]
                print(f"last: {last_page}")

finally:
    # Selalu tutup browser setelah selesai
    driver.quit()



# # for a in soupcnn.select('a[href*="page="]'):
# for a in soupcnn.select('a'):
#     print(a)
```
