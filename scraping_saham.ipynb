```python
import csv
import time
import pandas as pd

import os
```

```python
# !pip install yfinance
import yfinance as yf
```

```python
#rentang data
startdate = pd.to_datetime("2015-01-01")
enddate = pd.to_datetime("2025-08-31")

folder = 'data_saham'

# fungsi scraping sahamnya
def collectdata(x):
    try:
        history = yf.Ticker(x)
        df = history.history(start=startdate, end=enddate, auto_adjust=True)
        df.reset_index(inplace=True)  # ubah date dari index jdi kolom
        df['Date'] = pd.to_datetime(df['Date'])  # ubah "Date" menjadi datetime
        df['Date'] = df['Date'].iloc[0].date() #hpus yang date
        return df
    except Exception as e:
        print(f'kendala scrapping {x} | {e}')
        return ''

# fungsi ngesave csv-nya
def csvsaver(x):
    try:
        csv_filename = f'{folder}/{x}_data.csv'
        df = collectdata(x)
        df.to_csv(csv_filename, index=False)
        return f'saham {x} rentang waktu {startdate} hingga {enddate} berhasil di save di {folder}!'
    except Exception as e:
        print(f'kendala save csv {x} | {e}')
        return ''
```
