```python
df = pd.read_csv(
    "cnbc_labeled_manual_raw.csv",
    sep="\t",
    encoding="latin1",
    engine="python"
)

df.head()
```

```python
df.columns
```

```python
# regex + clean function
URL_RE = re.compile(r"(https?://\S+|www\.\S+)", flags=re.IGNORECASE)
HTML_RE = re.compile(r"<[^>]+>")
WS_RE = re.compile(r"\s+")
SYMBOL_RE = re.compile(r"[^a-zA-Z0-9\s\.,!?\'\-]")

def clean_text(s: str) -> str:
    if pd.isna(s):
        return ""
    s = str(s)
    s = HTML_RE.sub(" ", s)
    s = URL_RE.sub(" ", s)
    s = SYMBOL_RE.sub(" ", s)
    s = s.lower()
    s = WS_RE.sub(" ", s).strip()
    return s

#cleaning title & content
df["title_clean"] = df["title"].apply(clean_text)
df["content_clean"] = df["content"].apply(clean_text)
```

```python
df.head()
```

```python
#collect needed data
df_final = df[
    ["date", "title_clean", "content_clean", "sentiment", "article_id"]
].copy()

df_final = df_final.rename(
    columns={
        "title_clean": "title",
        "content_clean": "content"
    }
)

# gabungin buat input IndoBERT
df_final["text"] = (
    df_final["title"].str.strip() + ". " +
    df_final["content"].str.strip()
).str.strip()

df_final.head()
```

```python
#normalized label
sentiment_map = {
    "positive": 1,
    "neutral": 0,
    "negative": -1
}

df_final["sentiment"] = df_final["sentiment"].str.lower().str.strip()
df_final["label"] = df_final["sentiment"].map(sentiment_map)

#bikin date jdi datetime
df_final["date"] = pd.to_datetime(
    df_final["date"],
    format="%m/%d/%Y",
    errors="coerce"
)


# buang sentiment string
df_final = df_final.drop(columns=["sentiment"])
```
